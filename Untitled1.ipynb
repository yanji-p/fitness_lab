```python
df['workout_type'].fillna(df['workout_type'].mode()[0], inplace=True)
```

```python
df = df[df['steps'] > 0]
```

```python
df = df[df['sleep_hours'] > 0]
```

```python
df = df[df['distance_km'] > 0]
```

```python
def remove_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    return df[(df[col] >= Q1 - 1.5*IQR) & (df[col] <= Q3 + 1.5*IQR)]

for col in ['steps', 'calories_burned', 'sleep_hours', 'heart_rate_avg']:
    df = remove_outliers(df, col)
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
mood_map = {'bad': 0, 'neutral': 1, 'good': 2}
df['mood'] = df['mood'].map(mood_map)
```

```python
df = pd.get_dummies(df, 
                    columns=['workout_type', 'weather_conditions', 'location'], 
                    drop_first=True)
```

```python
df['date'] = pd.to_datetime(df['date'])
df['day_of_week'] = df['date'].dt.dayofweek   # 0=Monday, 6=Sunday
df['month']       = df['date'].dt.month
df['is_weekend']  = df['date'].dt.dayofweek.isin([5, 6]).astype(int)
df = df.drop(columns=['date'])
```
